# Task 01

```python
def get_alternate_elements(input_list):
    return input_list[::2]

user_list = input("Enter elements of the list separated by space: ").split()
print("Alternate elements of the list are:", get_alternate_elements(user_list))
```

# Task 03

```python
user_list = []
for i in range(10):
    num = input(f"Enter number {i+1}: ")
    user_list.append(num)

largest = int(user_list[0])

for num in user_list[1:]:
    if int(num) > largest:
        largest = int(num)

print("The largest number in the list is:", largest)
```

# Task 06

```python
date_input = input("Enter a date in the format 'dd-mm-yyyy': ")
day, month, year = date_input.split("-")

month_names = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
month_name = month_names[int(month) - 1]

# Print the date in the desired format
print(f"{month_name} {day}, {year}")
```

# Task 09

```python
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]

result = [[matrix1[i][j] + matrix2[i][j] for j in range(len(matrix1[0]))] for i in range(len(matrix1))]
print("Resultant matrix:")
for row in result:
    print(row)
```

# Task 10

```python
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8],
    [9, 10],
    [11, 12]
]

result = [[sum(matrix1[i][k] * matrix2[k][j] for k in range(len(matrix2))) for j in range(len(matrix2[0]))] for i in range(len(matrix1))]
print("Resultant matrix:")
for row in result:
    print(row)
```
