# Task 01

```python
numbers = [10, 25, 14, 7, 32, 18, 9, 40]
total = 0

print("All numbers: ")
for number in numbers:
    print(number)
    total += number

print("Sum:", total)

print("Even numbers: ")
for number in numbers:
    if number % 2 == 0:
        print(number)
```

# Task 02

```python
student = {
    "name": "Muhammad",
    "age": 20,
    "course": "Applied Computing",
    "marks": [75, 82, 68]
}

print("Student Details:")

for key, value in student.items():
    print(key, ":", value)
    
total = 0

for mark in student["marks"]:
    total += mark

average = total / len(student["marks"])

print("Average:", average)

if average >= 50:
    print("Passed")
else:
    print("Failed")
```

# Task 03

```python
employees = [
    {
        "name": "Ali",
        "department": "IT",
        "salary": 60000
    },
    {
        "name": "Ahmed",
        "department": "HR",
        "salary": 55000
    },
    {
        "name": "Usman",
        "department": "Finance",
        "salary": 75000
    }
]

print("Employee Details:")

for employee in employees:
    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print()

highest_salary = employees[0]
total_salary = 0

for employee in employees:
    total_salary += employee["salary"]

    if employee["salary"] > highest_salary["salary"]:
        highest_salary = employee

print("Highest Salary Employee:", highest_salary["name"])
print("Highest Salary:", highest_salary["salary"])
print("Total Salary Expense:", total_salary)
```

# Task 04

```python
numbers = []

while True:
    number = int(input("Enter a number (-1 to stop): "))

    if number == -1:
        break

    numbers.append(number)

if len(numbers) > 0:
    print("Complete List:", numbers)
    print("Largest Number:", max(numbers))
    print("Smallest Number:", min(numbers))
else:
    print("No numbers were entered.")
```

# Task 05

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word Frequency:")

for word, count in frequency.items():
    print(word, ":", count)
```
