# Printing Name Father Name and DOB

```python
name = input("Enter your name :")
father_name = input("Enter your father name :")
dob = int(input("Enter your age"))

print("Your name :", name , "\n" , "Your father's name :" , father_name , "\n" , "Your date of birth :" , dob)
```

# Operators

```python
a = 10
b = 5
c = a + b  # Addition

d = a - b  # Subtraction

e = a * b  # Multiplication

f = a / b  # Division

g = a % b  # Modulus

h = a ** b  # Exponentiation

print("Addition:", c)
print("Subtraction:", d)
print("Multiplication:", e)
print("Division:", f)
print("Modulus:", g)
print("Exponentiation:", h)
```

# Small Tasks

```python
english_marks = 85
islamiat_marks = 90
maths_marks = 95
total_marks = 300
percentage = (english_marks + islamiat_marks + maths_marks) / total_marks * 100
print("Percentage:", percentage)
```

# Swaping Values without Third Variable

```python
a = 10
b = 5
print("Before swapping:")
print("a =", a)
print("b =", b)
a = a + b
b = a - b
a = a - b
print("After swapping:")
print("a =", a)
print("b =", b)
```

# Calculating Area & Circumference

```python
radius = 5
pi = 3.14159
area = pi * radius ** 2
circumference = 2 * pi * radius

print("Area of the circle:", area)
print("Circumference of the circle:", circumference)
```

# Calculating Discount Amount & Final Price

```python
price = float(input("Enter the price of the item: "))
discount_percentage = float(input("Enter the discount percentage: "))

discount_amount = price * (discount_percentage / 100)
final_price = price - discount_amount

print("Discount amount:", discount_amount)
print("Final price after discount:", final_price)
```
